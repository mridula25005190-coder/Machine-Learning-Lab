```python

import pandas as pd
from urllib.parse import urlparse
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score

# Sample dataset
data = {
    "url": [
        "https://www.google.com",
        "https://www.wikipedia.org",
        "https://www.amazon.com",
        "http://login-example.com/verify",
        "http://secure-login@abc.com",
        "http://free-prize-example.com/win"
    ],
    "label": [0, 0, 0, 1, 1, 1]
}

df = pd.DataFrame(data)

print("Original Dataset:")
print(df)

# Feature extraction
def extract_features(url):
    return [
        len(url),
        url.count("."),
        url.count("-"),
        sum(c.isdigit() for c in url),
        sum(c in "@?=&" for c in url),
        int(url.startswith("https")),
        url.count("/")
    ]

# Convert URLs into numerical features
X = df["url"].apply(extract_features).tolist()
y = df["label"]

print("\nExtracted Features:")
print(X)

# Split dataset into training and testing
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=42
)

# Train Logistic Regression model
model = LogisticRegression()
model.fit(X_train, y_train)

# Make predictions
y_pred = model.predict(X_test)

# Calculate accuracy
accuracy = accuracy_score(y_test, y_pred)

# Display results
print("\nPredicted Labels:", y_pred)
print("Actual Labels:", list(y_test))
print("Accuracy:", accuracy)

# Display URL classification
for url, prediction in zip(
    df["url"].iloc[y_test.index], y_pred
):
    if prediction == 0:
        print(url, "-> Safe URL")
    else:
        print(url, "-> Malicious URL")
```
